# Data Pre-processing (pipeline)

```python
import os
import sys

sys.path.append(os.path.abspath("../"))
from src.config import BASE_PATH
from src.preprocess import transform_data, export_data
import pandas as pd
import joblib
```

Load from memory

```python
df_hopkins = pd.read_parquet(
    BASE_PATH / "data/raw/hopkins_ORN_extra_clean.parquet",
).drop("ID", axis=1)
X = df_hopkins.drop("ORN", axis=1)
y = df_hopkins["ORN"]
base_pipeline = joblib.load(BASE_PATH / "data/pipelines/base_pipeline.joblib")
nomo_pipeline = joblib.load(BASE_PATH / "data/pipelines/nomo_pipeline.joblib")
```

```python
for col in df_hopkins:
    n_na = df_hopkins[col].isna().sum()
    if n_na > 0:
        print(col)
        print(n_na)
```

Transform

```python
base_data = transform_data(X, base_pipeline)
nomo_data = transform_data(X, nomo_pipeline)
```

Export

```python
export_data(
    base_data,
    y,
    exp_path=BASE_PATH / "data" / "processed" / "base",
)
export_data(
    nomo_data,
    y,
    exp_path=BASE_PATH / "data" / "processed" / "nomo",
)
```
